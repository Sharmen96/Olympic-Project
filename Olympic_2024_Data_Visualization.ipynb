## Athelete  Information

### No of Males and Female Athlete

```python
print(df['gender'].value_counts())
#print(df['gender'].unique()) for variable
```

```python
# Countplot under distribution plot
sns.countplot(df,x=df['gender'])
```

### Top 10 countries with highest number  of Athletes

```python
top = df['country'].value_counts().head(10).index
print(top)

# value_counts() will sort descendingly (automatically)
```

```python
# filter the original dataset based on above countries only using the index list
top_10 = df[df['country'].isin(top)]
print(top_10.head())
```

```python
plt.figure(figsize=(25,15))
sns.countplot(top_10, x=top_10['country'], hue= 'gender')
```

### Top 5 countries with highest no of male participants


```python
male = df[df['gender']== 'Male']
top_male = male['country']. value_counts().head(5).index
top_male_countries = male[male['country'].isin(top_male)]
print(top_male_countries)
```

```python
# the point is to get the row of data(filtered data) that satisfies the question/instruction
```

```python
sorted_countries = top_male[::-1] # sort ascendingly

print(sorted_countries)
```

```python
sns.countplot(top_male_countries, x=top_male_countries['country'], order= sorted_countries)
```

### Top 5 countries with highest no of female participants

```python
#female data only
female = df[df['gender']== 'Female']
# country index list
top_female = female['country']. value_counts().head(5).index
# filter data based on country index list( if the country column is in country index list)

top_female_countries = female[female['country'].isin(top_female)]
print(top_female_countries)
```

```python
sorted_countries1 = top_female[::-1] # sort ascendingly

print(sorted_countries1)
```

```python
sns.countplot(top_female_countries, x='country', order= sorted_countries1)
```

### Top 8 games with highest no of participants

## Athlete age analysis

```python
from datetime import datetime
df['birth_date'] = pd.to_datetime(df['birth_date'])
df['age'] = df['birth_date'].apply(lambda birth_date: (datetime.now()- birth_date).days // 365)

# double slash means i want to get the integer or the age without decimal after dividing with 365
```

```python
bins = [0, 18,25,35,45,55,65,75,85,100]
labels = ['0-18', '19-25','26-35','46-55','56-65','66-75','76-85','86-100']
sns.histplot(data=df, x='age', discrete =True,kde =True)
```
